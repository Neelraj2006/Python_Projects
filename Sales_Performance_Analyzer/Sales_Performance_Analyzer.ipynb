# Sales Performance Analyzer

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data={
    "Product":["Laptop","Phone","Tablet","Watch","Headphones","Speaker"],
    "Sales":[50000,30000,15000,10000,8000,12000],
    "Orders":[25,40,18,15,20,12]
}
df=pd.DataFrame(data)
```

```python
df
```

**Analysis**

***Highest and Lowest Selling Product***

```python
print("Highest selling product:\n", df.nlargest(1,"Sales")[["Product","Sales"]].to_string(index=False))
print("Lowest selling product:\n", df.nsmallest(1,"Sales")[["Product","Sales"]].to_string(index=False))
```

***Average Sales***

```python
print("Average Sales:",df["Sales"].mean())
```

***Total Revenue***

```python
print("Total Revenue:",df["Sales"].sum())
```

***Top 3 Products***

```python
print("Top 3 products:\n",df.nlargest(3,"Sales")[["Product","Sales"]])
```

***Bottom 3 Products***

```python
print("Bottom 3 products:\n",df.nsmallest(3,"Sales")[["Product","Sales"]])
```

***Sales Ranking***

```python
df["Rank"]=df["Sales"].rank(ascending=False, method="dense").astype(int)
print("Sales Ranking:\n",df[["Product","Sales","Rank"]].sort_values("Rank").to_string(index=False))
```

***Bar Chart***

```python
plt.bar(df["Product"],df["Sales"])
plt.title("Sales of Products")
plt.xlabel=("Products")
plt.ylabel=("Sales")
```

***Piechart***

```python
plt.title("Sales of Products")
plt.pie(df["Sales"],labels=df["Product"])
plt.show()
```

***Histogram***

```python
plt.title("Sales of Products")
plt.hist(df["Sales"])
plt.show()
```
