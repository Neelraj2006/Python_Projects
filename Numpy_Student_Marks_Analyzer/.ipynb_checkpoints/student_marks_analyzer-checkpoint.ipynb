# Student Mark Analyzer

```python
import numpy as np
```

```python
names=np.array(["Prashant","Tushar","Neelraj","Kevin","Ruturaj","Pushkar"])
marksarr=np.array([97,84,56,72,89,77])
```

**CLASS PERFORMANCE REPORT**

```python
print("Average marks:", marksarr.mean())
print("Highest marks:", marksarr.max())
ranker_student=np.argmax(marksarr)
print(f"\nTopper : {names[ranker_student]} ({marksarr[ranker_student]})")
print("Lowest marks:", marksarr.min())
lowest_mark_student=np.argmin(marksarr)
print(f"\nLowest Scorer: {names[lowest_mark_student]} ({marksarr[lowest_mark_student]})")
top_3_marks=np.sort(marksarr)[-3:][::-1]
print("Top 3 marks are:", top_3_marks)
pass_mark=60
results=np.where(marksarr>=pass_mark,"Pass","Fail")
print("Results are:", results)
```

**PASS-FAIL Analysis**

```python
passed_student=np.sum(marksarr>=pass_mark)
failed_student=np.sum(marksarr< pass_mark)

print("Students passed are:", passed_student)
print("Students failed are:", failed_student)

passed_percentage=(passed_student/len(marksarr))*100
print("Student Passed:",passed_percentage)
```

**Grade Distribution**

```python
for i in range (len(marksarr)):
    if marksarr[i]>=90:
        grade=" A"
    elif marksarr[i]>=80:
        grade=" B"
    elif marksarr[i]>=70:
        grade=" C"
    elif marksarr[i]>=60:
        grade=" D"
    else:
        grade=" F"

    print(f"{names[i]} : {marksarr[i]} -> Grade{grade}")
```

**Additional Statistics**

```python
print("Median Marks:", np.median(marksarr))
import matplotlib.pyplot as plt

plt.bar(names,marksarr)

plt.title("Student Marks Analysis")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.show()
```
