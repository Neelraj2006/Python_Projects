# Project: Student Performance Dashboard

```python
import numpy as np
import pandas as pd
```

```python
StudentInfo={
    "Name":["Neelraj","Ruturaj","Prashant","Harsh","Tushar","Shyam"],
    "Marks":[67,84,92,78,95,58],
    "Attendance":[95,88,72,81,98,65]
}
```

```python
df=pd.DataFrame(StudentInfo)
```

**Display Complete Student Data**

```python
df
```

```python
df.head()
```

**Statistical Analysis**

```python
print("MARKS ANALYSIS")
print("Average Marks:",df["Marks"].mean())
print("Highest Marks:",df["Marks"].max())
print("Lowest Marks:",df["Marks"].min())

print("\n")

print("ATTENDANCE ANALYSIS")
print("Average Attendance:",df["Attendance"].mean(), "%")
print("Highest Attendance:",df["Attendance"].max(),"%")
print("Lowest Attendance:",df["Attendance"].min(),"%")
```

**Topper's Identification**

```python
topper=df["Marks"].idxmax()
print("Topper of the class:")

print(f"Topper:{df.loc[topper,'Name']} ({df.loc[topper,'Marks']}marks)")
```

**Highest & Lowest Marks**

```python
highest_attendance=df["Attendance"].idxmax()
lowest_attendance=df["Attendance"].idxmin()

print("\nAttendance Analysis")

print(f"Highest Attendance : {df.loc[highest_attendance,'Name']} ({df.loc[highest_attendance,'Attendance']}%)")
print(f"Lowest Attendance : {df.loc[lowest_attendance,'Name']} ({df.loc[lowest_attendance,'Attendance']}%)")
```

**Student Scoring Above 80**

```python
print("\nStudent Scoring Above 80")
score=df[df["Marks"]>80]
print(score)
```

**Low Attendance Warning**

```python
print("\nLow Attendance Alert")

low_attendance=df[df["Attendance"]<75]
print(low_attendance)
```

**Grade Assignment**

```python
def calculate_grade(marks):
    if marks>=90:
        return "A"
    elif marks>=80:
        return "B"
    elif marks>=70:
        return "C"
    elif marks>=60:
        return "D"
    elif marks>=50:
        return "E"
    else:
        return "F"

df["Grade"]=df["Marks"].apply(calculate_grade)
print(df)
```

**Grade Distribution**

```python
print("\nGrade distribution")
print(df["Grade"].value_counts())
```

**Sort Student by Marks**

```python
print("\nStudents Sorted by Marks")
sorted_df=df.sort_values(by="Marks", ascending=False)
print(sorted_df)
```

**Top 3 students**

```python
print("\nTop 3 students")
top_3=df.sort_values(by="Marks", ascending=False).head(3)
print(top_3)
```

**Bottom 3 Students**

```python
print("\nBottom 3 students")
bottom_3=df.sort_values(by="Marks").head(3)
print(bottom_3)
```

**Pass Percentage**

```python
print("Pass Percentage")
passed_students=len(df[df["Marks"]>=40])
total_students=len(df)

pass_percentage=(passed_students/total_students)*100
print(f"\nPassing Percentage is: {pass_percentage:2f} %")
```
